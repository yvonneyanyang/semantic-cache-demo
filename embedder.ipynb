```python
# embedder.py
# Thin wrapper around Gemini embeddings.

import os
import google.generativeai as genai

EMBED_MODEL = "text-embedding-004"

class Embedder:
    def __init__(self, api_key: str | None = None):
        # Prefer an explicit key; otherwise read from environment variables.
        key = api_key or os.getenv("GOOGLE_API_KEY") or os.getenv("GEMINI_API_KEY")
        if not key:
            raise RuntimeError("Missing GOOGLE_API_KEY / GEMINI_API_KEY in environment.")
        genai.configure(api_key=key)

    def embed(self, text: str) -> list[float]:
        """Return an embedding vector for the given text."""
        out = genai.embed_content(model=EMBED_MODEL, content=text)
        return out["embedding"]

    def dim(self) -> int:
        """Probe the embedding dimensionality once."""
        return len(self.embed("probe"))
```
